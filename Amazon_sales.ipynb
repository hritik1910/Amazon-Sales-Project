```python
# df1['Category']=df1['Category'].astype(str)
data = df1['Category']
plt.figure(figsize=(10,5))
plt.hist(data, bins= 20, color='red', edgecolor= 'black')
plt.xticks(rotation=45)
plt.show()
```

```python
B2B_Check= df1['B2B'].value_counts()
myexplode= (0.7, 0)
plt.pie(B2B_Check, labels= B2B_Check, autopct= '%1.1f%%', explode= myexplode)
plt.legend()
plt.show()
```

```python
B2B_Check= df1['B2B'].value_counts()
myexplode= (0.7, 0)
plt.pie(B2B_Check, labels= B2B_Check.index, autopct= '%1.1f%%', explode= myexplode)
plt.legend()
plt.title('Buyer category')
plt.show()
```

```python
x1= df1['Category']
y1= df1['Size']
plt.scatter(x1,y1)
plt.xlabel('Category', fontweight='bold', color= 'darkorange')
plt.ylabel('Size', color= 'darkorange', fontweight='bold')
plt.yticks(rotation=45)
plt.show()
```
